```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_excel(r"C:\Users\kshal\Desktop\Student_Productivity.xlsx")
```

```python
data.isnull().sum()
```

```python
sns.set(style = "whitegrid")
```

```python
#1. Histogram – Productivity Score Distribution
plt.figure(figsize = (10, 6))
plt.hist(data['productivity_score'], bins = 30, color = "green")
plt.title("Productivity Score Distribution")
plt.xlabel("Productivity Score")
plt.ylabel("Frequency")
plt.savefig("productivity_score.png")
plt.show()
```

✅ Insight

The histogram shows that most students have a productivity score between 30 and 50. The distribution is slightly right-skewed, indicating that fewer students achieve very high productivity levels. This suggests that while most students maintain moderate productivity, only a small group performs at a very high level.

```python
# Bar Plot – Academic Level vs Students
plt.figure(figsize = (6, 4))
sns.countplot(x = "academic_level", data = data, color = "Skyblue")
plt.title("Student by Academic Level")
plt.xlabel("Academic Level")
plt.ylabel("Student")
plt.savefig("academic_level.png")
plt.show()
```

✅ Insight

The bar chart shows the distribution of students across different academic levels. Some academic levels have a higher number of students compared to others, indicating an uneven distribution. This may impact overall performance trends across levels.

```python
# Scatter Plot – Study Hours vs Exam Score
plt.figure(figsize=(10,6))
plt.scatter(data['study_hours'], data['exam_score'])
plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.savefig("study_hours.png")
plt.show()
```

✅ Insight

The scatter plot shows a positive relationship between study hours and exam scores. Students who study more tend to achieve higher exam scores. This indicates that study time plays a significant role in academic performance.

```python
# Box Plot – Productivity by Gender
plt.figure(figsize=(6,4))
sns.boxplot(x='gender', y='productivity_score', data=data)
plt.title("Productivity by Gender")
plt.savefig("gender.png")
plt.show()
```

✅ Insight

The box plot compares productivity scores across genders. The distribution appears similar, but slight differences in median or spread may indicate variation in productivity levels. Outliers suggest that some students perform significantly higher or lower than others.

```python
# Heatmap – Correlation Matrix 

plt.figure(figsize=(12,10))

corr = data.corr(numeric_only=True)

sns.heatmap(
    corr, annot=True, fmt=".2f",             
    cmap="coolwarm",
    linewidths=0.5,
    annot_kws={"size":8} 
    )
plt.title("Correlation Heatmap",  fontsize=16)
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.savefig("correlation_heatmap.png", dpi=300, bbox_inches='tight')
plt.show()
```

✅ Insight

The heatmap shows the correlation between different numerical features. Productivity score has a strong positive correlation with exam score, indicating that more productive students perform better academically. Burnout level has a negative correlation with both productivity and exam scores, suggesting that higher stress reduces performance. Sleep hours also negatively correlate with burnout, showing that proper rest helps reduce stress levels.

```python
# Pairplot – Feature Relationships

sns.pairplot(data[['study_hours','sleep_hours','productivity_score','exam_score']])
plt.savefig("working_hour.png")
plt.show()
```

✅ Insight

The pairplot provides a visual representation of relationships between multiple variables. It confirms that study hours and productivity score are positively related to exam performance, while burnout negatively impacts both productivity and results.

```python
# Burnout vs Productivity

plt.figure(figsize=(6,4))
sns.scatterplot(x='burnout_level', y='productivity_score', data=data, color = "green")
plt.title("Burnout vs Productivity")
plt.savefig("burnout_level.png")
plt.show()
```

✅ Insight

The scatter plot shows a negative relationship between burnout level and productivity score. As burnout increases, productivity tends to decrease. This indicates that mental well-being plays an important role in maintaining productivity.

```python
plt.figure(figsize=(6,4))
sns.histplot(data['exam_score'], bins=30)
plt.title("Exam Score Distribution")
plt.savefig("exam_score.png")
plt.show()
```

✅ Insight

The exam score distribution shows that most students score in the mid-range, with fewer students achieving very high or very low scores. This indicates a moderate overall academic performance among students.
